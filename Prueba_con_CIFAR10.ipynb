```python
def tensor_to_rgb(image):

    """
    image:
        [3, H, W]

    Pasa de [-1,1] a [0,1]
    y de [C,H,W] a [H,W,C].
    """

    image = (
        image.detach()
        .cpu()
    )

    image = (
        image + 1
    ) / 2

    image = image.clamp(
        0,
        1
    )

    image = image.permute(
        1,
        2,
        0
    )

    return image


plt.figure(
    figsize=(8, 8)
)


for i in range(16):

    plt.subplot(
        4,
        4,
        i + 1
    )

    plt.imshow(
        tensor_to_rgb(
            x0[i]
        )
    )

    plt.axis("off")


plt.suptitle(
    "Muestras originales de CIFAR-10"
)

plt.tight_layout()

plt.show()
```

# ============================================================
# 4. SCHEDULE DE DIFUSIÓN
#
# Conservamos exactamente el mismo esquema usado con MNIST.
# ============================================================

```python
T = 1000

beta_start = 1e-4
beta_end = 0.02


betas = torch.linspace(
    beta_start,
    beta_end,
    T,
    device=device
)


alphas = (
    1.0
    -
    betas
)


alpha_bars = torch.cumprod(
    alphas,
    dim=0
)


print(
    "beta inicial:",
    betas[0].item()
)

print(
    "beta final:",
    betas[-1].item()
)

print(
    "alpha_bar final:",
    alpha_bars[-1].item()
)
```

# ============================================================
# 5. PROCESO FORWARD
# ============================================================

```python
def forward_diffusion_sample(
    x0,
    t
):

    """
    x0:
        [B, 3, 32, 32]

    t:
        [B]

    xt =
        sqrt(alpha_bar_t) x0
        +
        sqrt(1-alpha_bar_t) epsilon
    """

    noise = torch.randn_like(
        x0
    )


    alpha_bar_t = (
        alpha_bars[t]
    )


    alpha_bar_t = (
        alpha_bar_t[
            :,
            None,
            None,
            None
        ]
    )


    xt = (

        torch.sqrt(
            alpha_bar_t
        )
        *
        x0

        +

        torch.sqrt(
            1.0
            -
            alpha_bar_t
        )
        *
        noise
    )


    return xt, noise
```

# ============================================================
# 7. TIME EMBEDDING
#
# Igual al modelo de MNIST.
# ============================================================

```python
class SinusoidalTimeEmbedding(
    nn.Module
):

    def __init__(
        self,
        dim
    ):

        super().__init__()

        self.dim = dim


    def forward(
        self,
        t
    ):

        half_dim = (
            self.dim // 2
        )


        frequencies = (
            math.log(10000)
            /
            (
                half_dim - 1
            )
        )


        frequencies = torch.exp(

            torch.arange(
                half_dim,
                device=t.device
            )

            *

            -frequencies
        )


        angles = (

            t.float()[
                :,
                None
            ]

            *

            frequencies[
                None,
                :
            ]
        )


        embedding = torch.cat(

            [
                angles.sin(),
                angles.cos()
            ],

            dim=1
        )


        return embedding
```

# ============================================================
# 8. RESIDUAL BLOCK
#
# Igual al modelo de MNIST.
# ============================================================

```python
class ResBlock(
    nn.Module
):

    def __init__(
        self,
        in_channels,
        out_channels,
        time_dim
    ):

        super().__init__()


        self.conv1 = nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=3,
            padding=1
        )


        self.norm1 = nn.GroupNorm(
            8,
            out_channels
        )


        self.conv2 = nn.Conv2d(
            out_channels,
            out_channels,
            kernel_size=3,
            padding=1
        )


        self.norm2 = nn.GroupNorm(
            8,
            out_channels
        )


        self.time_mlp = nn.Sequential(

            nn.SiLU(),

            nn.Linear(
                time_dim,
                out_channels
            )
        )


        if (
            in_channels
            !=
            out_channels
        ):

            self.residual = nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=1
            )

        else:

            self.residual = (
                nn.Identity()
            )


    def forward(
        self,
        x,
        t_emb
    ):

        h = self.conv1(
            x
        )

        h = self.norm1(
            h
        )

        h = F.silu(
            h
        )


        time_information = (
            self.time_mlp(
                t_emb
            )
        )


        time_information = (
            time_information[
                :,
                :,
                None,
                None
            ]
        )


        h = (
            h
            +
            time_information
        )


        h = self.conv2(
            h
        )


        h = self.norm2(
            h
        )


        h = (
            h
            +
            self.residual(
                x
            )
        )


        return F.silu(
            h
        )
```

# ============================================================
# 9. U-NET RGB
#
# CAMBIOS IMPORTANTES:
#
# entrada:
#     1 canal -> 3 canales
#
# salida:
#     1 canal -> 3 canales
#
# El resto se conserva.
# ============================================================

```python
class SimpleUNet(
    nn.Module
):

    def __init__(
        self,
        time_dim=64
    ):

        super().__init__()


        # ----------------------------------------------------
        # Embedding temporal
        # ----------------------------------------------------

        self.time_embedding = (
            nn.Sequential(

                SinusoidalTimeEmbedding(
                    time_dim
                ),

                nn.Linear(
                    time_dim,
                    time_dim
                ),

                nn.SiLU(),

                nn.Linear(
                    time_dim,
                    time_dim
                )
            )
        )


        # ----------------------------------------------------
        # ENTRADA RGB
        #
        # CAMBIO:
        #
        # MNIST:
        #     1 -> 64
        #
        # CIFAR:
        #     3 -> 64
        # ----------------------------------------------------

        self.input_conv = nn.Conv2d(
            3,
            64,
            kernel_size=3,
            padding=1
        )


        # ----------------------------------------------------
        # Encoder
        # 32 x 32
        # ----------------------------------------------------

        self.block1 = ResBlock(
            64,
            64,
            time_dim
        )


        # 32 -> 16
        self.down1 = nn.Conv2d(
            64,
            128,
            kernel_size=4,
            stride=2,
            padding=1
        )


        self.block2 = ResBlock(
            128,
            128,
            time_dim
        )


        # 16 -> 8
        self.down2 = nn.Conv2d(
            128,
            256,
            kernel_size=4,
            stride=2,
            padding=1
        )


        # ----------------------------------------------------
        # Bottleneck
        # 8 x 8
        # ----------------------------------------------------

        self.middle = ResBlock(
            256,
            256,
            time_dim
        )


        # ----------------------------------------------------
        # Decoder
        # ----------------------------------------------------

        # 8 -> 16
        self.up1 = (
            nn.ConvTranspose2d(
                256,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            )
        )


        self.block3 = ResBlock(
            256,
            128,
            time_dim
        )


        # 16 -> 32
        self.up2 = (
            nn.ConvTranspose2d(
                128,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            )
        )


        self.block4 = ResBlock(
            128,
            64,
            time_dim
        )


        # ----------------------------------------------------
        # SALIDA RGB
        #
        # CAMBIO:
        #
        # MNIST:
        #     64 -> 1
        #
        # CIFAR:
        #     64 -> 3
        #
        # La red predice ruido RGB.
        # ----------------------------------------------------

        self.output_conv = nn.Conv2d(
            64,
            3,
            kernel_size=1
        )


    def forward(
        self,
        x,
        t
    ):

        # Embedding temporal
        t_emb = (
            self.time_embedding(
                t
            )
        )


        # Entrada
        x = self.input_conv(
            x
        )


        # Encoder
        skip1 = self.block1(
            x,
            t_emb
        )


        x = self.down1(
            skip1
        )


        skip2 = self.block2(
            x,
            t_emb
        )


        x = self.down2(
            skip2
        )


        # Bottleneck
        x = self.middle(
            x,
            t_emb
        )


        # Decoder
        x = self.up1(
            x
        )


        x = torch.cat(
            [
                x,
                skip2
            ],
            dim=1
        )


        x = self.block3(
            x,
            t_emb
        )


        x = self.up2(
            x
        )


        x = torch.cat(
            [
                x,
                skip1
            ],
            dim=1
        )


        x = self.block4(
            x,
            t_emb
        )


        return self.output_conv(
            x
        )
```

```python
model = SimpleUNet().to(
    device
)


n_parameters = sum(
    p.numel()
    for p in model.parameters()
)


print(
    "Número de parámetros:",
    f"{n_parameters:,}"
)


# ============================================================
# 11. PRUEBA DE UNA PASADA
#
# Antes de entrenar verificamos:
#
# x0              [B,3,32,32]
# xt              [B,3,32,32]
# epsilon         [B,3,32,32]
# epsilon_theta   [B,3,32,32]
# ============================================================

x0_test, _ = next(
    iter(loader)
)

x0_test = x0_test.to(
    device
)


t_test = torch.randint(
    0,
    T,
    (x0_test.shape[0],),
    device=device
)


xt_test, noise_test = (
    forward_diffusion_sample(
        x0_test,
        t_test
    )
)


with torch.no_grad():

    predicted_noise_test = model(
        xt_test,
        t_test
    )


print(
    "x0:",
    x0_test.shape
)

print(
    "xt:",
    xt_test.shape
)

print(
    "Ruido real:",
    noise_test.shape
)

print(
    "Ruido predicho:",
    predicted_noise_test.shape
)


test_loss = F.mse_loss(
    predicted_noise_test,
    noise_test
)


print(
    "Loss antes de entrenar:",
    test_loss.item()
)
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-4
)


EPOCHS = 50


epoch_losses = []


Path(
    "modelos"
).mkdir(
    exist_ok=True
)


Path(
    "Img CIFAR10"
).mkdir(
    exist_ok=True
)


for epoch in range(
    EPOCHS
):

    model.train()

    total_loss = 0.0


    for x0, _ in loader:

        x0 = x0.to(
            device,
            non_blocking=True
        )


        # Tiempo aleatorio para cada imagen
        t = torch.randint(
            0,
            T,
            (x0.shape[0],),
            device=device
        )


        # Forward
        xt, noise = (
            forward_diffusion_sample(
                x0,
                t
            )
        )


        # Predicción de ruido
        predicted_noise = model(
            xt,
            t
        )


        # MSE
        loss = F.mse_loss(
            predicted_noise,
            noise
        )


        optimizer.zero_grad()

        loss.backward()

        optimizer.step()


        total_loss += (
            loss.item()
        )


    average_loss = (
        total_loss
        /
        len(loader)
    )


    epoch_losses.append(
        average_loss
    )


    print(
        f"Época "
        f"{epoch + 1:2d}/"
        f"{EPOCHS} | "
        f"Pérdida promedio: "
        f"{average_loss:.6f}"
    )


    # --------------------------------------------------------
    # CHECKPOINT DE SEGURIDAD
    #
    # Sobrescribimos después de cada época.
    # Así no pierdes todo si se termina el job.
    # --------------------------------------------------------

    torch.save(

        model.state_dict(),

        "modelos/"
        "cifar10_diffusion_model.pt"
    )


print(
    "Entrenamiento terminado."
)
```

# ============================================================
# 15. PROCESO REVERSE
#
# Mismo algoritmo usado en MNIST.
# ============================================================

```python
@torch.no_grad()
def sample_reverse(
    model,
    n_samples=16
):

    model.eval()


    # --------------------------------------------------------
    # AHORA el ruido tiene 3 canales y resolución 32 x 32
    # --------------------------------------------------------

    x = torch.randn(
        n_samples,
        3,
        32,
        32,
        device=device
    )


    snapshots = {}


    save_times = [
        999,
        750,
        500,
        250,
        0
    ]


    for timestep in reversed(
        range(T)
    ):

        t = torch.full(
            (n_samples,),
            timestep,
            device=device,
            dtype=torch.long
        )


        # Predicción del ruido RGB
        predicted_noise = model(
            x,
            t
        )


        beta_t = betas[
            timestep
        ]


        alpha_t = alphas[
            timestep
        ]


        alpha_bar_t = (
            alpha_bars[
                timestep
            ]
        )


        # ----------------------------------------------------
        # Media del reverse
        # ----------------------------------------------------

        mean = (

            1
            /
            torch.sqrt(
                alpha_t
            )

        ) * (

            x

            -

            beta_t
            /
            torch.sqrt(
                1
                -
                alpha_bar_t
            )
            *
            predicted_noise
        )


        if timestep > 0:

            alpha_bar_previous = (
                alpha_bars[
                    timestep - 1
                ]
            )


            posterior_variance = (

                beta_t

                *

                (
                    1
                    -
                    alpha_bar_previous
                )

                /

                (
                    1
                    -
                    alpha_bar_t
                )
            )


            z = torch.randn_like(
                x
            )


            x = (

                mean

                +

                torch.sqrt(
                    posterior_variance
                )
                *
                z
            )


        else:

            x = mean


        if timestep in save_times:

            snapshots[
                timestep
            ] = (

                x.detach()
                .cpu()
            )


    return x, snapshots
```

```python

torch.manual_seed(
    42
)


if torch.cuda.is_available():

    torch.cuda.manual_seed_all(
        42
    )


generated_images, snapshots = (
    sample_reverse(
        model,
        n_samples=16
    )
)


# ============================================================
# 17. VISUALIZAR IMÁGENES GENERADAS
# ============================================================

plt.figure(
    figsize=(8, 8)
)


for i in range(16):

    plt.subplot(
        4,
        4,
        i + 1
    )


    image = tensor_to_rgb(
        generated_images[i]
    )


    plt.imshow(
        image
    )


    plt.axis(
        "off"
    )


plt.suptitle(
    "Muestras generadas por el DDPM sobre CIFAR-10"
)


plt.tight_layout()


plt.savefig(
    "Img CIFAR10/"
    "generated_cifar10.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 18. VISUALIZAR UNA TRAYECTORIA REVERSE
# ============================================================

times_to_show = [
    999,
    750,
    500,
    250,
    0
]


plt.figure(
    figsize=(14, 3)
)


for i, timestep in enumerate(
    times_to_show
):

    image = snapshots[
        timestep
    ][0]


    image = tensor_to_rgb(
        image
    )


    plt.subplot(
        1,
        len(times_to_show),
        i + 1
    )


    plt.imshow(
        image
    )


    plt.title(
        f"t={timestep}"
    )


    plt.axis(
        "off"
    )


plt.suptitle(
    "Proceso reverse del DDPM sobre CIFAR-10"
)


plt.tight_layout()


plt.savefig(
    "Img CIFAR10/"
    "reverse_process_cifar10.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Proceso CIFAR-10 terminado."
)
```

Reverse con diagnóstico de Score

```python
# ============================================================
# DIAGNÓSTICO DEL SCORE DURANTE EL REVERSE DE CIFAR-10
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

Path("Img CIFAR10").mkdir(exist_ok=True)

model.eval()


# Tiempos que queremos observar
save_times = [
    999,
    750,
    500,
    250,
    50,
    0
]


@torch.no_grad()
def reverse_with_score_diagnostics(
    model,
    n_samples=1
):

    model.eval()

    # Empezamos desde ruido gaussiano RGB
    x = torch.randn(
        n_samples,
        3,
        32,
        32,
        device=device
    )

    diagnostics = []

    # Para analizar cómo cambia la magnitud del score
    score_history = []


    # ========================================================
    # PROCESO REVERSE
    # ========================================================

    for timestep in reversed(range(T)):

        t_tensor = torch.full(
            (n_samples,),
            timestep,
            device=device,
            dtype=torch.long
        )


        beta_t = betas[timestep]

        alpha_t = alphas[timestep]

        alpha_bar_t = alpha_bars[timestep]


        # ====================================================
        # 1. LA U-NET PREDICE EL RUIDO RGB
        # ====================================================

        epsilon_pred = model(
            x,
            t_tensor
        )


        # ====================================================
        # 2. SCORE
        #
        # s_theta(x_t,t)
        #
        #          - epsilon_theta(x_t,t)
        # = --------------------------------
        #          sqrt(1 - alpha_bar_t)
        #
        # Shape:
        #
        # [B, 3, 32, 32]
        # ====================================================

        score_pred = (
            -epsilon_pred
            /
            torch.sqrt(
                1.0 - alpha_bar_t
            )
        )


        # ====================================================
        # 3. ESTIMACIÓN INSTANTÁNEA DE x_0
        # ====================================================

        x0_hat = (
            x
            -
            torch.sqrt(
                1.0 - alpha_bar_t
            )
            *
            epsilon_pred
        ) / torch.sqrt(
            alpha_bar_t
        )


        # ====================================================
        # 4. MAGNITUD DEL SCORE POR PÍXEL
        #
        # Cada píxel tiene:
        #
        # (s_R, s_G, s_B)
        #
        # Tomamos:
        #
        # sqrt(s_R² + s_G² + s_B²)
        # ====================================================

        score_magnitude = torch.linalg.vector_norm(
            score_pred,
            dim=1
        )


        mean_score_magnitude = (
            score_magnitude
            .mean()
            .item()
        )


        score_history.append(
            (
                timestep,
                mean_score_magnitude
            )
        )


        # ====================================================
        # 5. GUARDAR DIAGNÓSTICOS
        # ====================================================

        if timestep in save_times:

            diagnostics.append({

                "t":
                    timestep,

                "xt":
                    x.detach()
                    .cpu()
                    .clone(),

                "epsilon_pred":
                    epsilon_pred.detach()
                    .cpu()
                    .clone(),

                "score_pred":
                    score_pred.detach()
                    .cpu()
                    .clone(),

                "x0_hat":
                    x0_hat.detach()
                    .cpu()
                    .clone()
            })


        # ====================================================
        # 6. PASO REVERSE DDPM
        # ====================================================

        mean = (
            1.0
            /
            torch.sqrt(
                alpha_t
            )
        ) * (
            x
            -
            beta_t
            /
            torch.sqrt(
                1.0 - alpha_bar_t
            )
            *
            epsilon_pred
        )


        if timestep > 0:

            alpha_bar_previous = (
                alpha_bars[
                    timestep - 1
                ]
            )


            posterior_variance = (
                beta_t
                *
                (
                    1.0
                    -
                    alpha_bar_previous
                )
                /
                (
                    1.0
                    -
                    alpha_bar_t
                )
            )


            z = torch.randn_like(x)


            x = (
                mean
                +
                torch.sqrt(
                    posterior_variance
                )
                *
                z
            )


        else:

            x = mean


    diagnostics = sorted(
        diagnostics,
        key=lambda d: d["t"],
        reverse=True
    )


    return (
        x.detach().cpu(),
        diagnostics,
        score_history
    )
```

```python
# ============================================================
# FUNCIONES DE VISUALIZACIÓN
# ============================================================

def rgb_display(tensor):

    """
    Tensor [3,H,W] en escala aproximadamente [-1,1]
    -> imagen RGB [0,1].
    """

    image = (
        tensor
        .squeeze(0)
        .detach()
        .cpu()
    )

    image = (
        image + 1
    ) / 2

    image = image.clamp(
        0,
        1
    )

    image = image.permute(
        1,
        2,
        0
    )

    return image.numpy()



def magnitude_map(tensor):

    """
    tensor:
        [1,3,H,W]

    Devuelve:
        sqrt(R² + G² + B²)
    """

    tensor = (
        tensor
        .squeeze(0)
    )

    magnitude = torch.linalg.vector_norm(
        tensor,
        dim=0
    )

    return magnitude.numpy()



def show_magnitude(
    ax,
    tensor,
    title
):

    mag = magnitude_map(
        tensor
    )

    # Escala robusta para no dejar que unos pocos
    # valores extremos dominen toda la figura
    vmax = np.percentile(
        mag,
        99
    )

    if vmax < 1e-8:
        vmax = 1.0

    im = ax.imshow(
        mag,
        cmap="magma",
        vmin=0,
        vmax=vmax
    )

    ax.set_title(
        title,
        fontsize=10
    )

    ax.axis("off")

    return im


# ============================================================
# FIGURA
# ============================================================

n_rows = len(
    diagnostics
)

fig, axes = plt.subplots(
    n_rows,
    4,
    figsize=(
        12,
        2.8 * n_rows
    )
)


for i, result in enumerate(
    diagnostics
):

    timestep = result["t"]


    # --------------------------------------------------------
    # 1. x_t
    # --------------------------------------------------------

    axes[i, 0].imshow(
        rgb_display(
            result["xt"]
        )
    )

    axes[i, 0].set_title(
        rf"$x_t$, $t={timestep}$"
    )

    axes[i, 0].axis(
        "off"
    )


    # --------------------------------------------------------
    # 2. Magnitud del ruido predicho
    # --------------------------------------------------------

    eps_mag = magnitude_map(
        result["epsilon_pred"]
    )

    mean_eps = eps_mag.mean()


    show_magnitude(
        axes[i, 1],
        result["epsilon_pred"],
        rf"$||\epsilon_\theta||_2$"
        + "\n"
        + rf"media={mean_eps:.2f}"
    )


    # --------------------------------------------------------
    # 3. Magnitud del score
    # --------------------------------------------------------

    score_mag = magnitude_map(
        result["score_pred"]
    )

    mean_score = score_mag.mean()


    show_magnitude(
        axes[i, 2],
        result["score_pred"],
        rf"$||s_\theta||_2$"
        + "\n"
        + rf"media={mean_score:.2f}"
    )


    # --------------------------------------------------------
    # 4. x0_hat
    # --------------------------------------------------------

    x0_hat_display = (
        result["x0_hat"]
        .clamp(
            -1,
            1
        )
    )


    axes[i, 3].imshow(
        rgb_display(
            x0_hat_display
        )
    )


    axes[i, 3].set_title(
        r"$\hat{x}_0$"
    )


    axes[i, 3].axis(
        "off"
    )


plt.suptitle(
    "Del ruido al score durante la generación en CIFAR-10",
    fontsize=16,
    y=1.01
)


plt.tight_layout()


plt.savefig(
    "Img CIFAR10/"
    "cifar10_score_diagnostics.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()
```
